# 🧪 Maguru AI Model Explorer & Benchmark Notebook
Notebook ini digunakan untuk menguji ketersediaan model OpenRouter / Z.AI, memeriksa model gratis yang aktif, serta melakukan pengujian langsung dengan LangChain & LangGraph.

```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI

# Load .env file dari root directory maguru-model
load_dotenv('../.env')

openrouter_key = os.getenv('OPENROUTER_API_KEY')
openrouter_base = os.getenv('OPENROUTER_API_BASE', 'https://openrouter.ai/api/v1')
current_model = os.getenv('OPENROUTER_MODEL', '').strip()

print(f"🔑 OpenRouter Key loaded: {openrouter_key[:10]}..." if openrouter_key else "❌ OpenRouter Key not found")
print(f"🤖 Current Model in .env: '{current_model}'")
```

## 1. Cek Model Gratis (Free Tier) yang Sedang Aktif di OpenRouter

```python
# Ambil daftar model dari OpenRouter API secara real-time
resp = requests.get('https://openrouter.ai/api/v1/models')

if resp.status_code == 200:
    all_models = resp.json().get('data', [])
    free_models = [m for m in all_models if ':free' in m['id']]
    print(f"✅ Ditemukan {len(free_models)} model gratis yang terdaftar di OpenRouter:\n")
    for i, m in enumerate(free_models, 1):
        context_len = m.get('context_length', 'N/A')
        print(f"{i}. {m['id']} (Context: {context_len} tokens)")
else:
    print(f"❌ Gagal mengambil model: {resp.status_code} - {resp.text}")
```

## 2. Pengujian Live Model OpenRouter

```python
client = OpenAI(
    api_key=openrouter_key,
    base_url=openrouter_base
)

def test_model(model_name: str, prompt: str = "Jelaskan konsep variabel dalam 1 kalimat bahasa Indonesia"):
    print(f"\n⏳ Testing Model: {model_name} ...")
    try:
        response = client.chat.completions.create(
            model=model_name,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=100,
            temperature=0.7
        )
        content = response.choices[0].message.content
        print(f"✅ [SUCCESS] {model_name}:\n{content}")
        return True
    except Exception as e:
        print(f"❌ [FAILED] {model_name} -> Error: {e}")
        return False

# Uji coba beberapa kandidat model gratis
test_candidates = [
    "google/gemma-4-31b-it:free",
    "z-ai/glm-5.2:free",
    "openai/gpt-oss-20b:free",
    "nvidia/nemotron-3-nano-30b-a3b:free"
]

for model in test_candidates:
    test_model(model)
```
