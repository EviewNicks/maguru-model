## 3. Load Test Data

```python
# Load session content
session_content = load_session_content("python_basics", "module_1", "session_1_1")

if session_content is None:
    print("Warning: Session content not found, using fallback")
    session_content = """# Session 1.1: Pengenalan Variabel

## Tujuan Pembelajaran
- Memahami apa itu variabel
- Belajar cara membuat variabel di Python
- Mempraktikkan aturan penamaan variabel

## Konsep
Variabel adalah wadah untuk menyimpan nilai data.
"""
session_title = "Session 1.1: Pengenalan Variabel"

print(f"Session: {session_title}")
print(f"Content length: {len(session_content)} characters")
print("\\n--- First 500 chars ---")
print(session_content)
```

## 4. Test Cases

### Test 1: Question Without Chat History

```python
print("Test 1: First Question (No History)")
print("="*50)

question = "Apa itu variabel?"
chat_history = []

result = qa_chatbot.answer_question(
    question=question,
    session_title=session_title,
    session_content=session_content,
    chat_history=chat_history
)

print(f"\nQuestion: {question}")
print(f"\nResponse:\n{result}")
```

### Test 2: Question With Chat History

```python
print("Test 2: Follow-up Question (With History)")
print("="*50)

question = "Bagaimana cara membuat variabel yang benar?"
chat_history = [
    {"role": "student", "content": "Apa itu variabel?"},
    {"role": "ai", "content": "Variabel adalah wadah untuk menyimpan nilai data di Python."}
]

result = qa_chatbot.answer_question(
    question=question,
    session_title=session_title,
    session_content=session_content,
    chat_history=chat_history
)

print(f"\nChat History:")
for msg in chat_history:
    role = "Siswa" if msg["role"] == "student" else "AI"
    print(f"  {role}: {msg['content']}")

print(f"\nQuestion: {question}")
print(f"\nResponse:\n{result}")
```

### Test 3: Technical Question

```python
print("Test 3: Technical Question")
print("="*50)

question = "Kenapa nama variabel tidak boleh dimulai dengan angka?"

result = qa_chatbot.answer_question(
    question=question,
    session_title=session_title,
    session_content=session_content,
    chat_history=[]
)

print(f"\nQuestion: {question}")
print(f"\nResponse:\n{result}")
```

### Test 4: Multiple Questions

```python
print("Test 4: Multiple Questions")
print("="*50)

questions = [
    "Apa saja aturan penamaan variabel?",
    "Apa perbedaan huruf besar dan kecil dalam nama variabel?",
    "Bisakah saya menggunakan kata kunci Python sebagai nama variabel?"
]

for i, q in enumerate(questions, 1):
    result = qa_chatbot.answer_question(
        question=q,
        session_title=session_title,
        session_content=session_content,
        chat_history=[]
    )
    print(f"\nQ{i}: {q}")
    print(f"A: {result}...")
```

## 5. Analysis

```python
print("Analysis Summary")
print("="*50)

# Check response quality indicators
test_questions = [
    "Apa itu variabel?",
    "Bagaimana cara membuat variabel?",
    "Apa aturan penamaan?"
]

print("\nQuality Checks:")
for q in test_questions:
    result = qa_chatbot.answer_question(
        question=q,
        session_title=session_title,
        session_content=session_content,
        chat_history=[]
    )
    
    # Check for Indonesian
    indo_words = ["adalah", "untuk", "dengan", "yang", "ini", "itu"]
    has_indo = any(word in result.lower() for word in indo_words)
    
    # Check length
    length_ok = 50 < len(result) < 1000
    
    status = "OK" if (has_indo and length_ok) else "CHECK"
    print(f"  [{status}] '{q[:30]}...' - Length: {len(result)}, Indonesian: {has_indo}")
```
