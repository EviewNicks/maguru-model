## 4. Analysis

```python
print("Analysis Summary")
print("="*50)

# Test multiple code snippets
test_codes = [
    ("x = 5", "Variable"),
    ("print('hello')", "Print"),
    ("nama = input()", "Input"),
]

print("\nQuality Checks:")
for code, label in test_codes:
    result = explain_code.explain_code(code)
    
    # Check for Indonesian
    indo_words = ["adalah", "untuk", "dengan", "akan", "ini", "fungsi", "variabel"]
    has_indo = any(word in result.lower() for word in indo_words)
    
    # Check for explanation elements
    explains_what = any(word in result.lower() for word in ["melakukan", "fungsi", "untuk", "mencetak"])
    
    status = "OK" if (has_indo and explains_what) else "CHECK"
    print(f"  [{status}] {label}: Indonesian={has_indo}, Explains={explains_what}, Length={len(result)}")
```

```python
print("Response Quality:")

# Check if explanations follow the expected format
code = '''
nama = "Budi"
print(f"Halo {nama}")
'''
result = explain_code.explain_code(code)

# Expected elements from prompt
checks = {
    "Line-by-line": any(word in result.lower() for word in ["baris", "line", "setiap"]),
    "Why it works": any(word in result.lower() for word in ["karena", "mengapa", "sehingga"]),
    "Common mistakes": any(word in result.lower() for word in ["kesalahan", "hati-hati", "jangan", "hindari"]),
    "Simple language": len(result.split()) > 20  # Substantial explanation
}

for check, passed in checks.items():
    status = "OK" if passed else "CHECK"
    print(f"  [{status}] {check}")
```
